```python
features = ['Open', 'High', 'Low', 'Close', 'Volume']

plt.subplots(figsize=(20,10))

for i, col in enumerate(features):
  plt.subplot(2,3,i+1)
  sns.distplot(df[col])
plt.show()
```

```python
splitted = df['Date'].str.split('/', expand=True)

df['month'] = splitted[0].astype(int)
df['day'] = splitted[1].astype(int)
df['year'] = splitted[2].astype(int)
```

```python
df['Date'] = pd.to_datetime(df['Date'])

df['day'] = df['Date'].dt.day
df['month'] = df['Date'].dt.month
df['year'] = df['Date'].dt.year

df.head()
```

```python
df['day_of_week'] = df['Date'].dt.dayofweek
df['day_of_year'] = df['Date'].dt.dayofyear
df['quarter'] = df['Date'].dt.quarter
df['month_name'] = df['Date'].dt.month_name()
```

```python
df
```

```python
data_grouped = df.groupby('year')[['Open', 'High', 'Low', 'Close']].mean()

fig, axes = plt.subplots(2, 2, figsize=(20, 10))

for ax, col in zip(axes.flatten(), ['Open', 'High', 'Low', 'Close']):
    data_grouped[col].plot.bar(ax=ax)
    ax.set_title(f'Yearly Average {col} Price')
    ax.set_xlabel('Year')
    ax.set_ylabel('Average Stock Price')

plt.tight_layout()
plt.show()
```
